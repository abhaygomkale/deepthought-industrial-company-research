```python
import pandas as pd

# load both csvs
df22 = pd.read_csv("/content/google.csv")
df3 = pd.read_csv("/content/google (1).csv")

# combine them
df2 = pd.concat([df22, df3], ignore_index=True)

# preview
print(df2.head())

# total rows
print("Total Rows:", df2.shape[0])
```

```python
df2 = df2.dropna(how='all')
```

```python
df2.columns = [

    # Google Maps business listing URL
    'google_maps_link',

    # Company / business name
    'company_name',

    # Google rating
    'rating',

    # Number of reviews
    'review_count',

    # Business type/category
    'business_type',

    # separator dot
    'separator_1',

    # Address
    'address',

    # Open/closed status
    'business_status',

    # Opening timing
    'opening_time',

    # separator dot
    'separator_2',

    # Phone number
    'phone_number',

    # Official website
    'website',

    # Website button icon
    'website_icon',

    # Website text
    'website_text',

    # Direction button icon
    'direction_icon',

    # Direction text
    'direction_text',

    # Business category image/icon
    'category_icon',

    # Review snippet / customer highlight
    'review_snippet',

    # Extra metadata
    'extra_info_1',

    # Extra metadata
    'extra_info_2'
]
```

```python
clean_df2 = df2[[
    'company_name',
    'business_type',
    'address',
    'phone_number',
    'website',
    'rating',
    'review_count',
    'review_snippet',
    'opening_time',
    'google_maps_link',
]]
```

```python
clean_df2 = clean_df2.dropna(subset=['company_name', 'website'])
```

```python
clean_df2.head()
```

```python
clean_df2.shape
```

```python
keywords = [
    'manufacturer',
    'automation',
    'instrumentation',
    'industrial',
    'electronics'
]

pattern = '|'.join(keywords)

filtered_df = clean_df2[
    clean_df2['business_type']
    .astype(str)
    .str.lower()
    .str.contains(pattern, na=False)
]
```

```python
bad_keywords = [
    'dealer',
    'supplier',
    'shop',
    'repair',
    'cctv',
    'service'
]

bad_pattern = '|'.join(bad_keywords)

clean_df2 = filtered_df[
    ~filtered_df['business_type']
    .astype(str)
    .str.lower()
    .str.contains(bad_pattern, na=False)
]
```

```python
clean_df2.shape
```

```python
clean_df2 = clean_df2.drop_duplicates(
    subset='company_name'
)
```

```python
clean_df2['review_count_num'] = (
    clean_df2['review_count']
    .astype(str)
    .str.replace(r'[^0-9]', '', regex=True)
)

clean_df2['review_count_num'] = pd.to_numeric(
    clean_df2['review_count_num'],
    errors='coerce'
)
```

```python
top_df = clean_df2.sort_values(
    by=['review_count_num', 'rating'],
    ascending=False
)
```

```python
top_100 = top_df.head(100)
```

```python
top_100.to_csv("top_100_companies.csv", index=False)
```
